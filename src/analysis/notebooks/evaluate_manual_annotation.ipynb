# Evaluation of manual annotation for Semi-Automatic method
This notebook shows some overall statistics from the manual annotation.

```python
# import libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# read annotations
annotator_1 = pd.read_csv("../../../data/processed/manual_annotation/annotator_1.csv")
annotator_2 = pd.read_csv("../../../data/processed/manual_annotation/annotator_2.csv")

# remove rows with empty file names
annotator_1 = annotator_1[annotator_1["File / Case ID"].notna()]
annotator_2 = annotator_2[annotator_2["File / Case ID"].notna()]

# concatenate df's
annotations = pd.concat([annotator_1, annotator_2])

# remove one duplicate. after checking the data the mask was accepted. therefore the row with reject as decision is removed.
# because it's first we can just keep the first entry
annotations = annotations.drop_duplicates(subset="File / Case ID", keep="first")

print(annotations["5. Final \r\nDecision"].value_counts())
```
